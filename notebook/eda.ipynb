```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
%matplotlib inline
```

```python
# CSV file semicolon (;) separated hai
df = pd.read_csv("../data/e-shop clothing 2008.csv", sep=";")
print("\n📌 Dataset Loaded Successfully")
```

```python
# Dataset Shape & Columns
print("\n📊 Dataset Shape (Rows, Columns):")
print(df.shape)

print("\n📋 Column Names:")
print(df.columns.tolist())
```

```python
# Missing Values Check
print("\n❓ Missing Values Count:")
print(df.isnull().sum())
```

```python
# Conversion Column Banana
# Rule: Agar user page 4 ya 5 tak gaya = Conversion
df["conversion"] = df["page"].apply(lambda x: 1 if x >= 4 else 0)

print("\n✅ Conversion Column Created")

# Conversion distribution
print("\n📈 Conversion Value Counts:")
print(df["conversion"].value_counts())

# Conversion rate
conversion_rate = (df["conversion"].sum() / len(df)) * 100
print(f"\n📊 Conversion Rate: {conversion_rate:.2f}%")
```

```python
# Correlation analysis with conversion
# Sirf numeric columns lena
numeric_df = df.select_dtypes(include=["int64", "float64"])

# Conversion ke saath correlation
correlation_df = pd.DataFrame(
    numeric_df.corr()["conversion"]
)

correlation_df.columns = ["correlation"]
correlation_df = correlation_df.sort_values("correlation", ascending=False)

print("\n📊 Feature Correlation with Conversion:")
print(correlation_df)
```

```python
# Summary and recommendations
print("\n🎯 EDA Completed Successfully")
print("\n📝 Recommendations for Model Training:")
print("1. 'page' feature has highest correlation (0.63)")
print("2. 'order' feature has moderate correlation (0.03)")
print("3. Consider using: page, order, price, price 2, month, day, country")
print("4. Remove: year (constant value), session ID (too many unique values)")
```
